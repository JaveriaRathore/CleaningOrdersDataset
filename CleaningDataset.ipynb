```python
# Issues in Dataset
# Missing values in:
# Age
# City
# OrderDate
# Amount
# Duplicate rows:
# 1 duplicate row
# Wrong data types:
# Age contains text value "twenty"
# Amount contains values like "$2000"
# OrderDate stored as object/string
```

```python
before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Rows Before:", before)
print("Rows After:", after)
print("Duplicates Removed:", before-after)
```

```python
df.loc[:,"Age"] = pd.to_numeric(df["Age"], errors="coerce")
```

```python
df.loc[:,"Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df
```

```python
df.loc[:, "Amount"] = df["Amount"].str.replace("$", "", regex=False)
```

```python
df.loc[:, "Amount"] = pd.to_numeric(df["Amount"], errors="coerce")
```

```python
df.loc[:, "Amount"] = df["Amount"].fillna(df["Amount"].median())
```

```python
df
```

```python
df.loc[:, "City"] = df["City"].fillna(df["City"].mode()[0])
```

```python
df["City"] = df["City"].str.title()
```

```python
df
```

```python
df.loc[:, "OrderDate"] = pd.to_datetime(
    df["OrderDate"],
    errors="coerce",
    dayfirst=True
)
```

```python
df.loc[:, "OrderDate"] = df["OrderDate"].fillna(df["OrderDate"].mode()[0])
```

```python
df
```

```python
print("Rows Before:", len(df))
print("Duplicates:", df.duplicated().sum())
```

```python
df.to_csv("cleaned_orders.csv", index=False)

print("Cleaned dataset exported successfully!")
```
